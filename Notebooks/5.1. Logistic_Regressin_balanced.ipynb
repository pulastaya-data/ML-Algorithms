### Logistic Regression Implementation 
- on synthetically generated data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

RANDOM_STATE = 15
```

```python
from sklearn.datasets import make_classification

X, y = make_classification(n_samples=1200, n_features=9, n_classes=2, random_state=RANDOM_STATE)
```

```python
df = pd.DataFrame(X, columns=[f'feature{i}' for i in range(1, 10)])

df['target'] = y
```

```python
df.head()
```

```python
X = df.iloc[:, :-1]
X
```

```python
y = df.iloc[:, -1]
y
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=RANDOM_STATE)
```

```python
print(f"X_train Shape: {X_train.shape}")
print(f"X_test Shape: {X_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression

logistic = LogisticRegression()
logistic.fit(X_train, y_train)
```

#### Why not fit_tranform() X_test?
Since make_classification generates data that is already fairly normalized, you might not notice a big difference here, but in real-world data, skipping scaling with Logistic Regression can seriously hurt performance.

```python
y_pred = logistic.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

score=accuracy_score(y_test, y_pred)
print("Accuracy Score =", score)
print("\n")
print("Classification Report:\n", classification_report(y_test ,y_pred))
print("Confusion Matrix:\n" ,confusion_matrix(y_test, y_pred))
```

#### Hyperparameter Tuning and Crossvalidation

GridSearchCV

```python
model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
c_values = [100, 10, 1, 0.1, 0.01]
solver = ['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params = dict(penalty=penalty, C=c_values, solver=solver)
```

```python
params
```

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold()
```

```python
from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(estimator=model,
                   param_grid=params,
                   scoring='accuracy',
                   cv=cv,
                   n_jobs=-1)
```

```python
print(grid)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred2 = grid.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred2)
print("Accuracy Score =", score)
print("Classification Score:\n", classification_report(y_test, y_pred2))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred2))
```

- Randomized Search CV

```python
from sklearn.model_selection import RandomizedSearchCV

randomcv=RandomizedSearchCV(estimator=model, param_distributions=params, cv=5, scoring='accuracy')
```

```python
randomcv.fit(X_train,y_train)
```

```python
randomcv.best_score_
```

```python
randomcv.best_params_
```

```python
y_pred3=randomcv.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred3)
print("Accuracy Score =", score)
print("Classification Score:\n", classification_report(y_test, y_pred3))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred3))
```

#### Logistic Regression for multiclass classification

```python
X, y = make_classification(n_samples=1000, n_features=10, n_informative=3, n_classes=3, random_state=15)
```

```python
pd.DataFrame(X).head()
```

```python
X_train, X_test, y_train, y_test=train_test_split(X, y,
                                               test_size=0.25,
                                               random_state=42)
```

```python
from sklearn.multiclass import OneVsRestClassifier

model = OneVsRestClassifier(LogisticRegression())
```

```python
model.fit(X_train, y_train)
```
