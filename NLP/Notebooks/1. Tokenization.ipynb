## Tokenzation

`Tokenization` is the process of converting a `paragraph` or a `sentence` into `tokens`. If a paragraph is converted into sentences, each sentence is a token. If it is converted into words, each word is a token.

`Tokenization` is part of `text pre-processing`. Every word must eventually be converted into a `vector`, so the text first has to be broken into individual units.

```python
corpus = """Hello, I am Pulastaya. 
I have an interest in ML, NLP and DL. 
Yo! really its me."""
```

```python
print(corpus)
```

```python
corpus_2 = """This is Pulastaya's Git Hub repository. 
Its about Machine Learning."""
```

```python
print(corpus_2)
```

## Tokenization Techniques

#### 1. Sentence tokenization: sent_tokenize

- Converts a paragraph into sentences.
- Returns a list of sentences, which are the documents.
- It produced 4 sentences, split at the full stop, the exclamation mark and the end of the text.

```python
from nltk.tokenize import sent_tokenize

documents = sent_tokenize(corpus)
print(documents)
```

```python
documents_2 = sent_tokenize(corpus_2)
print(documents_2)
```

#### 2. Word Tokenization: word_tokenize

- Converts a paragraph or a sentence into words.

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(corpus)
print(words)
```

```python
words_2 = word_tokenize(corpus_2)
print(words_2)
```

- To tokenize sentence by sentence, loop over documents and apply word_tokenize to each sentence.

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
for sentence in documents_2:
    print(word_tokenize(sentence))
```

#### 3. Punctuation-based tokeniziation: wordpunct_tokenize

- Splits on every punctuation character.
- The aspstrophe in Pulastaya's is split off.

```python
from nltk.tokenize import wordpunct_tokenize

wordpuct = wordpunct_tokenize(corpus)
print(wordpuct)
```

```python
wordpuct_2 = wordpunct_tokenize(corpus_2)
print(wordpuct_2)
```

#### 4. Treebank tokenization: TreebankWordTokenizer

- A class: create an object first, then call .tokenize().
- A full stop is separated as its own token only at the very end of the text. A full stop in the middle stays attached to the previous word.

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

```python
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus_2)
```
