## Lemmatization

- `Lemmatization` technique is like stemming. The output we will get after lemmatization is called `lemma`, which is a root word rather than root stem, the output of stemming.
- After lemmatization, we will be getting a valid word that means the same thing.
- Example: eating, eats, eaten all reduce to eat.

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
'''
POS:
Noun-n
verb-v
adjective-a
adverb-r
'''

words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]

for word in words:
    print(word+"---->"+lemmatizer.lemmatize(word, pos='v'))
```

```python
lemmatizer.lemmatize("fairly",pos='v'),lemmatizer.lemmatize("sportingly")
```

Because the default is noun, a verb like going comes back unchanged. With pos="v" it becomes go.

```python
l1 = lemmatizer.lemmatize("going")
l2 = lemmatizer.lemmatize('going', pos='v')

print(f"Default: {l1}\nWith pos='v': {l2}")
```
