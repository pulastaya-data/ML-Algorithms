## Named Entity Recognition (NER)

#### What is named entity recognition?
- POS tagging says whether a word is a noun, pronoun, verb and so on. NER goes further and assigns entity tags saying what kind of real-world thing a word or phrase refers to.
- NER builds directly on POS tagging: words are tagged first, then grouped into entities.

**Example sentence:** "The Eiffel Tower was built from 1887 to 1889 by French engineer Gustave Eiffel, whose company specialized in building metal frameworks and structures."

| Entity type | Meaning | Example |
|---|---|---|
| **PERSON** | Names of people | Gustave Eiffel |
| **LOCATION / GPE** | Places (GPE = geo-political entity: countries, cities, nationalities) | Eiffel Tower (as a place), French |
| **ORGANIZATION** | Companies and institutions | Eiffel |
| **DATE / TIME** | Dates and times | 1887, 1889 |
| **MONEY** | Monetary values | "$1 million" (hypothetical) |

```python
sentence = "The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
import nltk

words = nltk.word_tokenize(sentence)
```

```python
words
```

```python
tag_elements = nltk.pos_tag(words)

tag_elements
```

```python
nltk.ne_chunk(tag_elements).draw()
```
